# Export a Table from a Feature Layer
- hosted in ArcGIS Enteprise 

- requires a Token

```python
import requests
import pandas as pd
```

```python
# REST query URL:
portal_rest = r'https://projects.udot.utah.gov/design/rest/services/PIN_19922_Digital_Delivery/FeatureServer/129/query'

# request params
params = {
    "where": "1=1",
    "outFields": "*",
    "returnGeometry": "false",
    "f": "json"
    }
```

```python
response = requests.get(portal_rest, params=params)
response.raise_for_status()

data = response.json()
```

```python
print(data)
```

```python
rows = [feature["attributes"] for feature in data.get("features", [])]

df = pd.DataFrame(rows)
```
