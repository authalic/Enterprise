# Item Copy/Update test
Can an item be copied (web map) and then re-linked to new data sources?

```python
from arcgis.gis import GIS
import pandas as pd
import helpers

# connect to GIS (Portal) using site admin creds
gis = helpers.get_gis('projects')
```

```python
def copy_portal_items(item_ids):  # list of item ids
    '''copies a list of portal items by ID, returns IDs of new copies'''

    new_ids = []

    for item_id in item_ids:
        print(f"Old item: {item_id}")

        old_item = gis.content.get(item_id)

        new_item = old_item.copy(
            old_item.title,
            old_item.tags,
            old_item.snippet,
            old_item.description
            )

        thumbnail_file = old_item.download_thumbnail(r'thumbnails')
        new_item.update(item_properties={}, thumbnail=thumbnail_file)

        new_ids.append(new_item.id)

        print(f"New item: {new_item.id}\n")

    return new_ids
```
