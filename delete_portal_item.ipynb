# Delete a Portal Item by its Item ID

- Why? Sometimes an Item gets stuck or orphaned in a Portal
- it can't be deleted from the Portal site itself
  - the Settings page might never load

```python
import helpers
```

```python
# available portal names for connection
portals = ['central', 'projects', 'regions', 'roads', 'uplan']

# connect to GIS (Portal) using site admin creds
gis = helpers.get_gis(portals[1])
```

```python
# use Item ID to get portal Item: SampleWorldCities

item_id = "5f9786308dd24eb9aecf5e9a105e48c3"
item = gis.content.get(item_id)
```

```python
# delete it

if item.delete():
    print("deleted")
else:
    print("failed to delete")
```
