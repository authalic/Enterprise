```python
import geopandas as gpd
```

```python
filepath_dbf = r"C:\Users\justinpjohnson\Downloads\shed_info_salt_cost.dbf"
filepath_csv = r"C:\Users\justinpjohnson\Downloads\shed_info_salt_cost.csv"
```
