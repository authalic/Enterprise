# Export a Table from a Feature Layer
- hosted in ArcGIS Enteprise 

```python
from arcgis.gis import GIS
import pandas as pd

from helpers import get_gis
```

```python
# Connect to the Enterprise portal.
# You can also use GIS("pro") if running from an authenticated ArcGIS Pro environment.
gis = get_gis('projects')
```

```python
# Get the feature layer item from Portal.
item = gis.content.get("c5eb6c4dee6a43c2bee5a6d2e7ec87da")

item
```

```python
# Choose the sublayer to export.
# Most feature layer items have one or more layers.
layer = item.layers[2]

layer
```

```python
# Query attributes only.
# return_geometry=False keeps the export focused on the attribute table.
sdf = layer.query(
    where="1=1",
    out_fields="*",
    return_geometry=False,
    as_df=True
)

# Optional: convert the spatially enabled dataframe to a plain pandas dataframe.
df = pd.DataFrame(sdf)
```
