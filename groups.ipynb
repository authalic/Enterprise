# Enterprise Groups


```python
from arcgis.gis import GIS
import helpers

# connect to GIS (Portal) using site admin creds
# gis = helpers.get_gis('central')
gis = helpers.get_gis('uplan')
```

## Find groups with hidden members

```python
groups = gis.groups.search()

for grp in groups:
    if grp.hiddenMembers:
        members = grp.get_members()
        print(f"{grp.title} -- {grp.homepage}")

        print(f"  owner:  {members['owner']}")
        print(f"  admins: {', '.join(members['admins'])}")
        print(f"  users:  {', '.join(members['users'])}")
        print('\n')
```
